# 11 — Docker and deployment validation

Build and run from the repository root. Docker packages the same application and pinned dependencies for repeatable local demonstrations.

```python
from pathlib import Path
print(Path('../Dockerfile').read_text())
print('\nBuild: docker build -t production-ml-monitoring .')
print('Run:   docker run --rm -p 5000:5000 production-ml-monitoring')
print('Open:  http://127.0.0.1:5000')
```

## Deployment notes

The demo writes prediction logs to a local CSV. For multi-container or cloud deployments, replace this with a durable database/object store, add authentication and request limits, and configure secrets through environment variables.
