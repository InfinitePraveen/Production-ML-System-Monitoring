# 05 — Reproducibility and model versioning

Model versions are immutable artifact files. A small registry JSON records the active version, metrics, and timestamp; production teams would usually back this with an artifact store and approval workflow.

```python
import json, platform
from datetime import datetime, timezone
from pathlib import Path
import sklearn
artifact_dir = Path('../artifacts'); artifact_dir.mkdir(exist_ok=True)
registry = {'project':'production-ml-system-monitoring','active_model':'model_v1.joblib','version':'1.0.0','created_at':datetime.now(timezone.utc).isoformat(),'python':platform.python_version(),'scikit_learn':sklearn.__version__,'status':'candidate'}
(artifact_dir/'model_registry.json').write_text(json.dumps(registry, indent=2))
registry
```

## Release checklist

1. Train and evaluate candidate model.
2. Save a new versioned artifact rather than overwriting the prior version.
3. Compare metrics against the current production model.
4. Update the registry only after review.
5. Keep the previous artifact available for rollback.
