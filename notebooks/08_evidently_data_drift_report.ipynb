# 08 — Evidently drift report (optional dependency)

Evidently's API evolves. This notebook uses the current `Report` / `DataDriftPreset` interface. If Evidently is not installed, use notebook 07's lightweight PSI calculation.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from pathlib import Path

data = load_breast_cancer(as_frame=True)
X = data.data
reference, current = train_test_split(X, test_size=0.25, random_state=7)
try:
    from evidently import Report
    from evidently.presets import DataDriftPreset
    report = Report([DataDriftPreset()])
    snapshot = report.run(current_dataset=current.reset_index(drop=True), reference_dataset=reference.reset_index(drop=True))
    Path('../artifacts').mkdir(exist_ok=True)
    snapshot.save_html('../artifacts/evidently_drift_report.html')
    print('Saved artifacts/evidently_drift_report.html')
except ImportError:
    print('Evidently is optional. Install requirements.txt to generate the report.')
except Exception as exc:
    print('Evidently report did not complete:', repr(exc))
    print('Check the installed Evidently version and its documentation.')
```
